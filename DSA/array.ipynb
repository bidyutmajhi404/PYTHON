# 2) Problem: Given an array of integers, return true if any value appears at least twice, and false if every element is distinct.

```python
def has_duplicate(arr):
    seen= set()
    for num in arr:
        if num in seen:
            return True
        seen.add(num)
    return False

nums=[1,2,3,2,4]
print(has_duplicate(nums))
```
