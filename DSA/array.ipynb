# 2) Problem: Given an array of integers, return true if any value appears at least twice, and false if every element is distinct.

```python
def has_duplicate(arr):
    seen= set()
    for num in arr:
        if num in seen:
            return True
        seen.add(num)
    return False

nums=[1,2,3,2,4]
print(has_duplicate(nums))
```

# 3) Find Maximum and Minimum Element in an Array

```python
def find_max_min(arr):
    if not arr:
        return None

    maximum = arr[0]
    minimum = arr[0]

    for num in arr:
        if num > maximum:
            maximum = num

        if num < minimum:
            minimum = num

    return maximum, minimum


nums = [3, 5, 1, 4, 2, 10]

max_val, min_val = find_max_min(nums)

print(f"Maximum => {max_val}")
print(f"Minimum => {min_val}")
```
