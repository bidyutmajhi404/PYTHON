## word Guessing Game

```python
import random 
```

```python
# user name 
name = input ("What is your name ?")
print("Good Luck!",name)
```

```python
# List of words
words=[
     'rainbow', 'computer', 'science', 'programming',
    'python', 'mathematics', 'player', 'condition',
    'reverse', 'water', 'board', 'geeks'
]
```

```python
# select a random word
word=random.choice(words)
print("\n Guess the Charcters")
# store guessed charcterers
guesses=''
# Number of attemps
turns=12
# Main game start 
while turns>0:
    failed=0

    for char in word:
        if char in guesses:
            print(char,end=" ")
        else:
            print("_",end=" ")
            failed+=1
    print()
    if failed==0:
        print("You win")
        print("the word is:",word)
        break

    # get user input 
    guess=input("Guess a charter : ").lower()

    if len(guess)!=1:
        print("Please enter a single charcter .")
        continue

    if guess in guesses:
        print("you already guessed that charcter .")
        continue
    guesses+=guess
     # Handle incorrect guesses
    if guess not in word:
        turns -= 1
        print("Wrong")
        print("You have", turns, "more guesses")

        # Check if user has lost
        if turns == 0:
            print("You Lose")
            print("The word was:", word)
```
