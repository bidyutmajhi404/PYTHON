### Read two integers and print their sum, difference, product, integer quotient, and remainder.
##### Example Input: 17 5
##### Example Output: sum=22, diff=12, product=85, quotient=3, remainder=2

```python
num1=int(input('Enter the value of num1'))
num2=int(input('Enter the value of num2'))
print(f'Sum: {num1+num2}')
print(f'Dif: {num1-num2}')
print(f'Product:{num1*num2}')
print(f'Quotient: {num1/num2}')
print(f'Remainder: {num1%num2}')
```
