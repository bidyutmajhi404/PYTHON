### Given a list of command-line style tokens such as `['--name', 'Asha', '--age', '30']`, build a dictionary of flag names and values without using argparse.


```python
tokens = ['--name', 'Asha', '--age', '30']

result = {}

i = 0
while i < len(tokens):
    key = tokens[i]        # --name
    value = tokens[i + 1]  # Asha

    result[key] = value

    i += 2   # jump to next pair

print(result)
```

### Build a mini REPL that repeatedly accepts `SET key value`, `GET key`, and `EXIT`, storing values in memory.

```python
store = {}

while True:
    command = input(">>")
    if command == "EXIT":
        break
    parts = command.split()
    if parts[0] == "SET":
        key = parts[1]
        value = parts[2]
        store[key]=value
        print('OK')
        
    elif parts[0] == "GET":
        key = parts[1]
        if key in store:
            print(store[key])
        else:
            print('NULL')
    else:
        print('Invalid Command')
        
```

### Implement a safe integer division prompt that keeps asking until the user enters valid integers and a nonzero denominato

### Write a parser that extracts key-value pairs from `a=1;b=2;c=hello` while ignoring empty segments and extra spaces.


```python
s = "a=1; b=2; ; c=hello ;"

result = {}

parts = s.split(";")

for part in parts:
    if part.strip() == "":
        continue

    if "=" in part:
        key, value = part.split("=")
        result[key.strip()] = value.strip()

print(result)
```
