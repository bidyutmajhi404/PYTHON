### Convert a temperature from Celsius to Fahrenheit and Kelvin, rounding each result to two decimal places.
##### Fahrenheit = (C × 9/5) + 32
##### Kelvin = C + 273.15

```python
celsius = int(input('Enter the temperature in celsius'))
fahrenheit = (celsius*9/5)+32
kelvin = celsius + 273.15

print(f'Fahrenheit: {fahrenheit} f')
print(f'Kelvin: {kelvin} k')
```

### Write a program that converts a total number of seconds into hours, minutes, and seconds.
##### Example Input: 3672
##### Example Output: 1 hour, 1 minute, 12 second
### Logic
##### Hours = total_seconds // 3600
##### Remaining seconds = total_seconds % 3600
##### Minutes = remaining // 60
##### Seconds = remaining % 60

```python
total_seconds=int(input('Enter value in seconds to convert in hours,minutes and seconds'))
hours = int (total_seconds/3600)
remaing = int(total_seconds%3600)
minutes = int(remaing/60)
seconds = int(remaing%60)

print(f'Hour: {hours}, Minutes: {minutes}, Seconds: {seconds}')
```

### Read three numbers and print the minimum, maximum, and arithmetic mean without using `min()` or `max()

```python
num1=int(input('Enter the value of num1'))
num2=int(input('Enter the value of num2'))
num3=int(input('Enter the value of num3'))

# find maximum value
if num1>=num2 and num1>=num3:
    maximum=num1
elif num2>=num1 and num2>=num3:
    maximum=num2
else:
    maximum=num3

# find mimimum value 
if num1<=num2 and num1<=num3:
    minimum=num1
elif num2<=num1 and num2<=num3:
    minimum=num2
else:
    minimum=num3

#mean value 
mean = (num1+num2+num3)/3

print(f'Maximum: {maximum}')
print(f'Minimum: {minimum}')
print(f'Mean: {mean}')
```
