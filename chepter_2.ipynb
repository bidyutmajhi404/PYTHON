### 32:Given a list of strings, remove all empty strings in place while preserving the order of non-empty items.


```python
strings = ["apple", "", "banana", "", "cherry", ""]

# Modify the same list (in-place)
strings[:] = [s for s in strings if s != ""]

print(strings)
```

### 34:Merge two dictionaries of inventory counts so that counts for common keys are added instead of overwritten.


```python
inv1 = {"apple": 10, "banana": 5, "orange": 8}
inv2 = {"banana": 7, "orange": 2, "grape": 4}

merged = inv1.copy()

for key, value in inv2.items():
    merged[key] = merged.get(key, 0) + value

print(merged)
```

### 35:Given two lists, return the elements that appear in both lists without duplicates, preserving the order from the first list.

```python
def common_ele(list1,list2):
    set2 = set(list2)
    seen = set()
    result = []

    for item in list1:
        if item in set2 and item not in seen:
            result.append(item)
            seen.add(item)
    return result


list1 = [1, 2, 2, 3, 4, 5]
list2 = [2, 3, 6, 2]

print(common_ele(list1, list2))
    
```
