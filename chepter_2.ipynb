### 1: Create a list from five integers and print the first, last, and middle elements.

```python
numbers = list (map (int,input('Enter five number seperate by space: ').split()))

print(f'First Element: {numbers[0]}')
print(f'Middle Element: {numbers[int(len(numbers)/2)]}')
print(f'Last Element: {numbers[-1]}')
```

### 2:Convert a tuple of student names into a list, add a new name, and convert it back to a tuple.

```python
# tupple of student 
students = ('Bidyut','Animesh','Sachin')
#convert is in list
student_list=list(students)
# add new name in the list
student_list.append('Raj')
#convert list to tuple
students = tuple(student_list)
print(students)
```

### 3:Given a list of numbers, create a set containing only the unique values and print the count of duplicates removed

```python
# inout list 
numbers = [1,2,2,2,2,3,4,4,4,4,5]
#unique value set
unique_numbers = set(numbers)
remove_numbers=len(numbers)-len(unique_numbers)
print(f'Number of duplicates remove: {remove_numbers}')
print(f'Set of unique numbers: {}')
```
