### 35:Given two lists, return the elements that appear in both lists without duplicates, preserving the order from the first list.

```python
def common_ele(list1,list2):
    set2 = set(list2)
    seen = set() # use to avoid duplicates
    result = []

    for item in list1:
        if item in set2 and item not in seen:
            result.append(item)
            seen.add(item)
    return result


list1 = [1, 2, 2, 3, 4, 5]
list2 = [2, 3, 6, 2]

print(common_ele(list1, list2))
    
```

### 38:Group a list of employee records by department using a dictionary where each key maps to a list of names.

```python
employees = [  
    {"name":"Bidyut","dept":"Backend devloper"},
    {"name":"Animesh","dept":"Cyber sequrity"},
    {"name":"Raj","dept":"Frontend Dev"},
    {"name":"Abhishek","dept":"Cloud"},
    {"name":"Joti","dept":"Frontend devloper"}
]
def group_by_department(employees):
    result = {}

    for emp in employees:
        dept = emp["dept"]
        name = emp["name"]

        if dept not in result:
            result[dept] = []

        result[dept].append(name)

    return result

print(group_by_department(employees))
```

### 39:Given a matrix represented as a list of lists, transpose it without using external libraries

```python
def transpose(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    
    result = []
    
    for i in range(cols):
        new_row = []
        for j in range(rows):
            new_row.append(matrix[j][i])
        result.append(new_row)
    
    return result

matrix = [
 [1, 2, 3],
 [4, 5, 6]
]
print(transpose(matrix))
```

### 40:Remove duplicates from a list of dictionaries based on the `id` field while keeping the first occurrence.

```python
data = [
    {"id":1,"name":"Animesh"},
    {"id":2,"name":"Bidyut"},
    {"id":3,"name":"Raj"},
    {"id":2,"name":"Gandu"},
    {"id":3,"name":"Bijoy"}
]
def remove_duplicates(data):
    seen = set()
    result = []
    #print(seen)
    for item in data:
        if item["id"] not in seen:
            seen.add(item["id"])
            result.append(item)
            #print(result)
    return result

print(remove_duplicates(data))
```

### 41:Create a frequency dictionary for all values in a list, then return the value with the highest frequency.

```python
def most_frequent(lst):
    freq = {}

    # count frequency
    for i in lst:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1

    # find max frequency
    max_value = None
    max_count = 0

    for key in freq:
        if freq[key] > max_count:
            max_count = freq[key]
            max_value = key

    return max_value

data=[1,2,3,4,1,2,3,2,2,2,2]
print(most_frequent(data))
```

```python
matrix = [[0] * 3] * 3

matrix[0][0] = 1
print(matrix)

# Explanation (aliasing issue)
#[[0] * 3] * 3 creates only one inner list
#That same list is reused 3 times
#So all rows are actually the same object

# fix the code 

matrix = [[0] * 3 for _ in range(3)]

matrix[0][0] = 1
print(matrix)
```

### 45:Implement a function that deeply flattens a nested list containing integers and lists of arbitrary depth

```python
def deep_flatten(lst):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.extend(deep_flatten(item))  # recursive call
        else:
            result.append(item)

    return result
```

```python
data = [1, [2, [3, 4], 5], 6, [7, [8, [9]]]]

print(deep_flatten(data))
```

### 46:Given two dictionaries representing sparse vectors, compute their dot product efficiently using only common keys

```python
def sparse_dot_product(vec1, vec2):
    # iterate over smaller dict for efficiency
    if len(vec1) > len(vec2):
        vec1, vec2 = vec2, vec1

    result = 0

    for key in vec1:
        if key in vec2:
            result += vec1[key] * vec2[key]

    return result
```

```python
v1 = {'a': 1, 'b': 2, 'c': 3}
v2 = {'b': 4, 'c': 5, 'd': 6}

print(sparse_dot_product(v1, v2))
```

### 47:Build an inverted index from a dictionary of document ids to word lists, mapping each word to the set of document ids containing it

```python
def build_inverted_index(docs):
    index = {}

    for doc_id, words in docs.items():
        for word in words:
            if word not in index:
                index[word] = set()
            index[word].add(doc_id)

    return index
```

```python
docs = {
    1: ["python", "code", "ai"],
    2: ["python", "ml"],
    3: ["ai", "ml", "data"]
}

index = build_inverted_index(docs)
print(index)
```
