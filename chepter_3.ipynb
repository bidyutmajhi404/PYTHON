### 51:Reverse a string without using `reversed()` and print the result.


```python
# use loop 
s = "hello"
reversed_s = ""

for ch in s:
    reversed_s = ch + reversed_s

print(reversed_s)
# Using slicing
s = "hello"
reversed_s = s[::-1]

print(reversed_s)
```

### 52:Count vowels, consonants, digits, and spaces in a given string.


```python
text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for ch in text:
    # Check vowel
    if ch.lower() in "aeiou":
        vowels += 1

    # Check consonant
    elif ch.isalpha():
        consonants += 1

    # Check digit
    elif ch.isdigit():
        digits += 1

    # Check space
    elif ch.isspace():
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```

### 53:Check whether a string is a palindrome after ignoring spaces and case.

```python
text = input("Enter a string: ")

# Remove spaces and convert to lowercase
cleaned = text.replace(" ", "").lower()

# Check palindrome
if cleaned == cleaned[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

### 54:Given a sentence, print each word on a new line after stripping punctuation at the ends.

```python
import string

sentence = input("Enter a sentence: ")

words = sentence.split()

for word in words:
    # Remove punctuation from both ends
    cleaned_word = word.strip(string.punctuation)

    print(cleaned_word)
```

## 57 Find the first non-repeating character in a string and return its index.


```python
def first_non_repeating(s):
    freq = {}

    # Count frequency
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    # Find first non-repeating character
    for i, ch in enumerate(s):
        if freq[ch] == 1:
            return i

    return -1


print(first_non_repeating("leetcode"))
print(first_non_repeating("loveleetcode"))
print(first_non_repeating("aabb"))
```
