### 51:Reverse a string without using `reversed()` and print the result.


```python
# use loop 
s = "hello"
reversed_s = ""

for ch in s:
    reversed_s = ch + reversed_s

print(reversed_s)
# Using slicing
s = "hello"
reversed_s = s[::-1]

print(reversed_s)
```

### 52:Count vowels, consonants, digits, and spaces in a given string.


```python
text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for ch in text:
    # Check vowel
    if ch.lower() in "aeiou":
        vowels += 1

    # Check consonant
    elif ch.isalpha():
        consonants += 1

    # Check digit
    elif ch.isdigit():
        digits += 1

    # Check space
    elif ch.isspace():
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```
