### 51:Reverse a string without using `reversed()` and print the result.


```python
# use loop 
s = "hello"
reversed_s = ""

for ch in s:
    reversed_s = ch + reversed_s

print(reversed_s)
# Using slicing
s = "hello"
reversed_s = s[::-1]

print(reversed_s)
```
